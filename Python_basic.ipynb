## 1. 숫자 (Number)

Python 은 정수형(`int`), 실수형(`float`), 복소수형(`complex`) 을 기본 제공합니다. 정수와 실수가 함께 연산되면 결과는 자동으로 실수형이 됩니다.

## 2. Boolean (논리 자료형)

`bool` 은 `True` / `False` 두 값만 갖는 자료형으로, 조건식의 결과나 흐름 제어에서 핵심적으로 쓰입니다. 내부적으로는 정수의 부분집합이며 `True == 1`, `False == 0` 입니다.

### 논리 연산자

| 연산자 | 의미 | 예 (`t=True, f=False`) | 결과 |
|---|---|---|---|
| `and` | 논리곱 (AND) | `t and f` | `False` |
| `or`  | 논리합 (OR) | `t or f` | `True` |
| `not` | 부정 (NOT) | `not t` | `False` |
| `==` / `!=` | 같다 / 다르다 (XOR 대용) | `t != f` | `True` |

### Truthy / Falsy
다음 값들은 자동으로 `False` 로 평가됩니다 (그 외는 `True`):
- `None`, `False`
- 숫자 `0`, `0.0`
- 빈 문자열 `""`, 빈 리스트 `[]`, 빈 딕셔너리 `{}`, 빈 튜플 `()`

````python
t = True
f = False
print(type(t)) # Prints "<class 'bool'>"
print(t and f) # Logical AND; prints "False"
print(t or f)  # Logical OR; prints "True"
print(not t)   # Logical NOT; prints "False"
print(t != f)  # Logical XOR; prints "True"
````

## 3. String (문자열)

문자열은 작은따옴표(`'...'`) 또는 큰따옴표(`"..."`) 로 둘러싸 만들며, 여러 줄은 삼중 따옴표(`'''...'''` / `"""..."""`) 로 표현합니다. **불변(immutable)** 자료형이라 한 번 만들면 개별 문자를 수정할 수 없고, 메서드는 항상 새 문자열을 돌려줍니다.

### 인덱싱 / 슬라이싱
```python
s = "Python"
s[0]      # 'P'      (앞에서 0 부터)
s[-1]     # 'n'      (음수는 뒤에서)
s[0:3]    # 'Pyt'    (start:end, end 미포함)
s[::2]    # 'Pto'    (step=2)
s[::-1]   # 'nohtyP' (역순)
```

### 자주 쓰는 메서드

| 메서드 | 설명 | 예 |
|---|---|---|
| `len(s)` | 길이 | `len("hello") → 5` |
| `s.upper()` / `s.lower()` | 대/소문자 변환 | `"hi".upper() → "HI"` |
| `s.capitalize()` / `s.title()` | 첫 글자 / 단어 첫 글자 대문자 | `"hi world".title() → "Hi World"` |
| `s.strip()` | 양쪽 공백 제거 | `"  hi ".strip() → "hi"` |
| `s.replace(old, new)` | 치환 | `"a-b".replace("-","_") → "a_b"` |
| `s.split(sep)` / `sep.join(list)` | 분리 / 결합 | `"a,b".split(",") → ["a","b"]` |
| `s.startswith(p)` / `s.endswith(p)` | 접두/접미사 검사 | `"file.csv".endswith(".csv") → True` |
| `s.center(w)` / `s.rjust(w)` / `s.ljust(w)` | 가운데/오른쪽/왼쪽 정렬 (폭 `w`) | `"hi".center(6) → "  hi  "` |

### 포매팅 (f-string, 권장)
```python
name, score = "Eunjoo", 95
print(f"{name} 의 점수는 {score} 입니다 ({score/100:.1%})")
# Eunjoo 의 점수는 95 입니다 (95.0%)
```

````python
s = "hello"
print(s)
print(len(s))           # string length
print(s.capitalize())   # Capitalize a string; prints "Hello"
print(s.upper())        # Convert a string to uppercase; prints "HELLO"
print(s.rjust(10))      # Right-justify a string, padding with spaces; prints "  hello"
print('1234567890')
print(s.center(10))     # Center a string, padding with spaces; prints " hello "
print(s.replace('l', '(ell)'))  # Replace all instances of one substring with another;
                                # prints "he(ell)(ell)o"
print('  world '.strip())  # Strip leading and trailing whitespace; prints "world"
````

## 4. 자료구조 — 리스트 / 딕셔너리 / 튜플 / 세트

여러 값을 하나로 묶는 4 가지 컴파운드(compound) 자료형. 무엇을 쓸지는 **순서가 필요한가? 키로 찾고 싶은가? 변경할 일이 있는가? 중복을 제거해야 하는가?** 로 결정합니다.

| 자료형 | 표기 | 순서 | 가변성 | 중복 | 키-값 | 대표 용도 |
|---|---|:-:|:-:|:-:|:-:|---|
| 리스트 (list)       | `[1, 2, 3]`        | O | **가변** | O | × | 순서 있는 모음, 자주 추가/삭제 |
| 튜플 (tuple)        | `(1, 2, 3)`        | O | 불변 | O | × | 변하면 안 되는 묶음, 다중 반환값 |
| 딕셔너리 (dict)     | `{'a': 1}`         | O¹ | **가변** | (키)× | O | 키로 빠른 조회, 매핑 |
| 세트 (set)          | `{1, 2, 3}`        | × | **가변** | × | × | 중복 제거, 집합 연산 |

¹ Python 3.7+ 에서 dict 은 삽입 순서를 보존합니다.

> 참고: <https://docs.python.org/ko/3/tutorial/datastructures.html>

### 4-1. 리스트 (list)

대괄호(`[]`) 안에 쉼표로 구분된 값들의 목록. 순서를 유지하는 **가변(mutable)** 시퀀스로, 서로 다른 자료형을 섞어 담을 수 있습니다.

| 동작 | 표현 | 비고 |
|---|---|---|
| 인덱싱 | `lst[i]` | 음수 인덱스 가능 (`lst[-1]` 마지막 원소) |
| 슬라이싱 | `lst[i:j:k]` | `j` 미포함 |
| 요소 변경 | `lst[i] = x` | 가변이라 가능 |
| 이어붙이기 | `lst1 + lst2` 또는 `lst1.extend(lst2)` | |
| 끝에 추가 | `lst.append(x)` | |
| 끝/특정 위치 제거 | `lst.pop()`, `lst.pop(i)` | 반환값 = 제거된 원소 |
| 정렬 | `lst.sort()` (in-place) / `sorted(lst)` (새 리스트) | |
| 길이 | `len(lst)` | |

````python
age = [12, 43, 29, 16, 25]
````

````python
age[0] = 31
````

### 4-2. 딕셔너리 (dict)

`{key: value}` 형태의 키-값 쌍 모음. 키는 **불변(hashable)** 자료형(문자열, 숫자, 튜플 등) 이어야 하며 중복될 수 없습니다. 값은 어떤 자료형이든 가능하고, 같은 키에 다시 대입하면 덮어씁니다.

| 동작 | 표현 |
|---|---|
| 생성 | `d = {'a': 1, 'b': 2}` 또는 `dict(a=1, b=2)` |
| 조회 | `d['a']` / 없을 때 기본값 → `d.get('z', 0)` |
| 추가/수정 | `d['c'] = 3` |
| 삭제 | `del d['a']` 또는 `d.pop('a')` |
| 키/값/쌍 순회 | `d.keys()`, `d.values()`, `d.items()` |
| 존재 확인 | `'a' in d` |

평균 조회 시간복잡도가 **O(1)** 이라 “키 → 값” 매핑이 필요할 때 가장 먼저 떠올려야 할 자료형입니다.

````python
tel_1 = {'mina': 4098, 'junho': 4139}
tel_2 = dict(tim=4139, david=4127, john=4098)
````

````python
print(tel_1, tel_2)
````

### 4-3. 튜플 (tuple)

쉼표로 구분된 여러 값으로 만들어지는 **불변(immutable)** 시퀀스. 보통 괄호(`()`) 로 감싸지만 괄호 없이 쉼표만 써도 튜플입니다. 인덱싱·슬라이싱은 가능하지만 개별 요소 대입은 불가합니다.

```python
memo = (12345, 54321, 'hello!')   # 괄호로 명시
point = 3, 4                       # 괄호 없이도 튜플
x, y = point                       # 언패킹 → x=3, y=4
```

| 특징 | 설명 |
|---|---|
| 불변 | `t[0] = ...` 불가 — 다시 만들어야 함 |
| 가변 요소는? | 튜플 안 리스트는 수정 가능 (`t[1][0] = 99` OK) |
| 다중 반환값 | 함수가 `return a, b` 하면 호출 측은 튜플로 받음 |
| 딕셔너리 키 | 튜플은 hashable 이라 dict 의 키로 사용 가능 (리스트는 불가) |

````python
memo = (12345, 54321, 'hello!')
````

````python
print(memo)
````

### 4-4. 세트 (set)

중복을 허용하지 않는 원소들의 모음. 순서가 없으며, 수학의 집합처럼 **합집합·교집합·차집합** 연산을 지원합니다. 빈 세트는 `set()` 으로 만들어야 합니다 (`{}` 는 빈 dict).

| 연산 | 연산자 | 메서드 |
|---|---|---|
| 합집합 (union) | `a \| b` | `a.union(b)` |
| 교집합 (intersection) | `a & b` | `a.intersection(b)` |
| 차집합 (difference) | `a - b` | `a.difference(b)` |
| 대칭차집합 (XOR) | `a ^ b` | `a.symmetric_difference(b)` |
| 원소 추가 / 제거 | — | `a.add(x)`, `a.remove(x)`, `a.discard(x)` |

> `remove(x)` 는 원소가 없으면 `KeyError`, `discard(x)` 는 조용히 무시합니다.
> 리스트의 중복 제거 패턴: `unique = list(set(my_list))` (순서는 보장되지 않음)

````python
 a = set([0,0,0,1,2,3])
 print(a)
````

````python
b = set()
b = {1,3,3,3,4,5,5,6}
print(b)
````

````python
inter1 = a & b
print(inter1)
inter2 = a.intersection(b)
print(inter2)
````

````python
union1 = a | b
print(union1)
union2 = a.union(b)
print(union2)
````

````python
sub1 = a - b
print(sub1)
sub2 = a.difference(b)
print(sub2)
````

````python

a.add(8)
print(a)
a.remove(8)
print(a)
a.discard(8) # 지우는 항목이 없어도 OK
a.remove(8) # 지우는 항목이 없으면 에러
````

## 5. 흐름 제어 — `if` / `for`

코드의 실행 순서를 조건이나 반복으로 제어합니다. Python 은 중괄호 대신 **들여쓰기(indent)** 로 블록을 구분합니다 (관례적으로 공백 4칸).

### 5-1. `if` / `elif` / `else`

조건을 평가해 참이면 해당 블록을 실행합니다. `elif` 는 0 개 이상, `else` 는 0 또는 1 개 사용 가능. 다른 언어의 `switch / case` 도 `if ... elif ...` 로 대체합니다.

```python
if 조건1:
    ...
elif 조건2:
    ...
else:
    ...
```

비교 연산자: `==`, `!=`, `<`, `<=`, `>`, `>=`
- 멤버십: `x in seq`, `x not in seq`
- 동일 객체: `a is b`, `a is not b` (값이 같은지가 아니라 **같은 객체인지**)
- 결합: `and`, `or`, `not`

> Python 3.10+ 에는 패턴 매칭(`match / case`) 도 추가됐지만, 입문 단계에서는 `if/elif` 로 충분합니다.

````python
x = int(input("Please enter your age: "))
````

````python
if x < 20:
    print('You are a teenager. Enjoy your life!')
elif x < 30:
    print('Work Hard for your future!')
elif x < 60:
    print('Take care of your family and you too')
else:
    print('Always be considerate of others')
````

### 5-2. `for`

C/Java 와 달리 “인덱스를 증가시키며” 도는 게 아니라 **시퀀스의 원소를 순서대로 꺼내** 도는 구조입니다 (foreach 스타일).

```python
for 원소 in 시퀀스:
    ...
```

자주 쓰는 패턴:

| 패턴 | 코드 |
|---|---|
| 인덱스 + 값 동시 | `for i, x in enumerate(seq):` |
| 두 시퀀스 동시 | `for a, b in zip(A, B):` |
| 정렬된 순서 | `for x in sorted(seq):` |
| 반대 순서 | `for x in reversed(seq):` |
| dict 의 key/value | `for k, v in d.items():` |

내부에서 흐름을 끊을 때:
- `break` — 반복 즉시 종료
- `continue` — 다음 반복으로 건너뜀
- `for ... else:` — `break` 없이 끝까지 다 돌았을 때만 `else` 실행

````python
animals = ['tiger', 'dog', 'elephant']
````

````python
for w in animals:
    print(w, len(w))
````

````python
for i, w in enumerate(animals):
    print(i, w)
````

#### `range()` — 숫자 시퀀스로 순회

`range(start, stop, step)` 은 **메모리에 리스트를 만들지 않고** 필요할 때마다 숫자를 생성하는 시퀀스 객체입니다 (게으른 평가). `stop` 은 포함되지 않습니다.

| 호출 | 생성되는 값 |
|---|---|
| `range(5)` | `0, 1, 2, 3, 4` |
| `range(2, 7)` | `2, 3, 4, 5, 6` |
| `range(0, 10, 3)` | `0, 3, 6, 9` |
| `range(10, 0, -1)` | `10, 9, 8, …, 1` |

리스트가 필요하면 `list(range(...))` 로 변환합니다.

````python
for i in range(5):
     print(i)
````

````python
for i in [0, 1, 2, 3, 4]:
    print(i)
````

````python
print(range(5))
````

````python
for i in range(0, 10, 3):
     print(i)
````

## 6. 함수

`def` 키워드로 함수를 정의합니다. 같은 코드를 여러 번 쓰지 않도록 묶어 재사용성과 가독성을 높이는 가장 기본적인 추상화 단위입니다.

```python
def 함수이름(매개변수1, 매개변수2=기본값):
    """함수 설명 (docstring)."""
    ...
    return 결과            # 생략하면 None 반환
```

### 인자 전달 방식

| 종류 | 정의 | 호출 예 |
|---|---|---|
| 위치 인자 (positional) | `def f(a, b):` | `f(1, 2)` |
| 기본 인자 (default) | `def f(a, b=2):` | `f(1)` → `b=2` |
| 키워드 인자 (keyword) | 정의는 같아도 호출 시 이름으로 | `f(a=1, b=2)` |
| 가변 위치 인자 | `def f(*args):` | `f(1, 2, 3)` → `args=(1,2,3)` |
| 가변 키워드 인자 | `def f(**kwargs):` | `f(x=1, y=2)` → `kwargs={'x':1,'y':2}` |

### 반환값
- `return` 이 없으면 `None` 반환
- 여러 값을 한 번에 반환하면 자동으로 튜플로 묶임 → `return a, b` ↔ 호출 측 `x, y = f()`

### 변수 범위 (scope)
함수 안에서의 이름 검색 순서: **L**ocal → **E**nclosing → **G**lobal → **B**uilt-in (LEGB). 함수 안에서 변수를 대입하면 별도 지정이 없는 한 모두 지역 변수가 됩니다.

````python
# 위치 인자 기본 인자
def add(a, b=2):
    c = a + b
    return c
````

````python
print(add(2))
````

````python
print(add(2,4))
````

````python
print(add(2,b=4))
````

## 7. 기타

### `with` 구문 (컨텍스트 매니저)

파일·소켓·DB 커넥션처럼 **사용 후 반드시 정리(close/해제) 해야 하는 리소스** 를 다룰 때 사용합니다. `with` 블록을 빠져나갈 때 — 정상 종료든 예외 발생이든 — 자동으로 정리 코드가 실행되어 리소스 누수를 막아 줍니다.

```python
# 권장 (with 구문)
with open("test.txt", "w") as f:
    f.write("hello")
# 블록을 벗어나면 f 가 자동으로 close 됨

# 비권장 (수동 close — 예외 시 누수 위험)
f = open("test.txt", "w")
f.write("hello")
f.close()
```

### 파일 모드 요약

| 모드 | 의미 |
|---|---|
| `'r'` | 읽기 (기본값, 파일 없으면 에러) |
| `'w'` | 쓰기 (있으면 덮어씀, 없으면 생성) |
| `'a'` | 이어쓰기 (append) |
| `'b'` | 바이너리 모드 — 다른 모드와 결합 (`'rb'`, `'wb'`) |
| `'x'` | 배타적 생성 (이미 있으면 에러) |
