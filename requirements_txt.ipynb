# requirements.txt 예시 — Ollama 기반 RAG

LangChain + Ollama 로 로컬 RAG (Retrieval-Augmented Generation) 파이프라인을 돌리기 위한 패키지 목록입니다. 본 셀은 실행 코드가 아니라 **`requirements.txt` 파일에 그대로 복사해 사용** 하기 위한 스니펫입니다.

## `requirements.txt` 란?
- 프로젝트가 의존하는 파이썬 패키지와 버전을 한 줄에 하나씩 적은 텍스트 파일
- `pip install -r requirements.txt` 한 번으로 동일한 환경 재현 가능
- 협업/배포 시 **버전 고정(pinning)** 으로 “내 컴퓨터에선 됐는데” 문제를 줄임

## 버전 표기 규칙

| 표기 | 의미 |
|---|---|
| `pkg==1.2.3` | 정확히 1.2.3 만 |
| `pkg>=1.2,<2.0` | 1.2 이상, 2.0 미만 |
| `pkg~=1.2.3` | 1.2.x (호환 버전) |
| `pkg` | 가장 최신 (재현성 ↓) |

## 이 목록의 핵심 패키지
- `langchain`, `langchain-core`, `langchain-community`, `langgraph` — RAG 체인/그래프 구성
- `langchain-openai`, `openai`, `tiktoken` — OpenAI 호환 LLM/임베딩
- `pypdf` — PDF 문서 로더
- `faiss-cpu` — 벡터 검색 인덱스
- `sentence-transformers` — 로컬 임베딩 모델
- `streamlit` — 데모 UI
